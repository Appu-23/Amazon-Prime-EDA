# ⚙️ Feature Engineering


## 1. Create Release Decade

```python
# Create Release Decade

titles_clean['release_decade'] = (titles_clean['release_year'] // 10) * 10

titles_clean[['release_year', 'release_decade']].head()
```

## 2. Create Primary Genre

```python
# Extract first genre

titles_clean['primary_genre'] = (
    titles_clean['genres']
    .str.strip("[]")
    .str.replace("'", "", regex=False)
    .str.split(",")
    .str[0]
)

titles_clean[['genres','primary_genre']].head()
```

## 3. Create Primary Production Country

```python
titles_clean['primary_country'] = (
    titles_clean['production_countries']
    .str.strip("[]")
    .str.replace("'", "", regex=False)
    .str.split(",")
    .str[0]
)

titles_clean[['production_countries','primary_country']].head()
```

## 4. Runtime Category

```python
def runtime_category(runtime):
    if runtime < 60:
        return "Short"
    elif runtime <= 120:
        return "Medium"
    else:
        return "Long"

titles_clean["runtime_category"] = titles_clean["runtime"].apply(runtime_category)

titles_clean[["runtime","runtime_category"]].head()
```

## 5. IMDb Rating Category

```python
def imdb_category(score):

    if pd.isna(score):
        return "Unknown"

    elif score < 5:
        return "Poor"

    elif score < 7:
        return "Average"

    elif score < 8.5:
        return "Good"

    else:
        return "Excellent"

titles_clean["imdb_category"] = titles_clean["imdb_score"].apply(imdb_category)

titles_clean[["imdb_score","imdb_category"]].head()
```

## Chart 21: Correlation Heatmap

```python
# Select numerical columns
numerical_cols = [
    "release_year",
    "runtime",
    "imdb_score",
    "imdb_votes",
    "tmdb_popularity",
    "tmdb_score",
    "content_age"
]

plt.figure(figsize=(10,8))

sns.heatmap(
    titles_clean[numerical_cols].corr(),
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()
```

## Chart 23: IMDb Score by Genre and Content Type

```python
top_genres = titles_clean["primary_genre"].value_counts().head(10).index

plt.figure(figsize=(14,6))

sns.boxplot(
    data=titles_clean[
        titles_clean["primary_genre"].isin(top_genres)
    ],
    x="primary_genre",
    y="imdb_score",
    hue="type"
)

plt.xticks(rotation=45)

plt.title("IMDb Score by Genre and Content Type")

plt.show()
```

## Chart 24: Average IMDb Score by Country

```python
country_rating = (
    titles_clean
    .groupby("primary_country")["imdb_score"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12,6))

country_rating.plot(
    kind="bar",
    color="teal"
)

plt.title("Top 10 Countries by Average IMDb Score")
plt.xlabel("Country")
plt.ylabel("Average IMDb Score")

plt.xticks(rotation=45)

plt.show()
```

### Why this chart?

It compares average IMDb scores across countries.

### Insight

Some countries consistently produce higher-rated content.

### Business Impact

Amazon can prioritize content acquisition from regions with strong audience ratings.

## Chart 25: Average TMDB Popularity by Genre

```python
genre_popularity = (
    titles_clean
    .groupby("primary_genre")["tmdb_popularity"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12,6))

genre_popularity.plot(
    kind="bar",
    color="orange"
)

plt.title("Top 10 Genres by Average TMDB Popularity")
plt.xlabel("Genre")
plt.ylabel("Average TMDB Popularity")

plt.xticks(rotation=45)

plt.show()
```

### Why this chart?
